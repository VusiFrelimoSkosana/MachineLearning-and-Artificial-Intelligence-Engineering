```python
#1. Base model is a core of pydantic and when used in a class it validates data type, convert it when posssible and raise an error if not possible

from pydantic import BaseModel

class user(BaseModel):
    name: str
    age: int

val= user(name="Vusi", age='25')

num=19

print(val.age)
print('addition:', num+ val.age)

#2.Inclid list and dictionary data type to 
from typing import List, Dict

class type(BaseModel):
    dataList: List[Dict]  #list of dictionaries
    dataStr: List[str]  #list of strings
    dataDict: Dict    #dictionary

print('__________________\n') 

listUser=[
    {"Name": "Alice", "Age": 25},
    {"Name": "Bob", "Age": 30}
]

dict={"Name": "Bob", "Age": 30}

listItems=["Vusi","Frelimo","Bheki"]

eval2=type(dataList=listUser, dataStr=listItems, dataDict=dict)

print(eval2.dataDict)
```

```python
#Define a car class

class Car:
    #Constructor method to initialize the instance
    def __init__(self,make,model,year):
        self.make=make #Attribte: make of the car
        self.model=model #Attribte: model of the car
        self.year=year #Attribte: year of the mabufacture
        
    # Method to display information about the car
    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")
        
    # Method to accelarate the car
    def accelarate(self,speed):
        print(f"{self.make} {self.model} is accelarating at {speed}")
        
#Create instances of the car
car1=Car('Toyota','Corolla',2022)
car2=Car('Ford','Mustang',2023)

# Access attributes and call methods

car1.display_info()
car2.accelarate(120)
```

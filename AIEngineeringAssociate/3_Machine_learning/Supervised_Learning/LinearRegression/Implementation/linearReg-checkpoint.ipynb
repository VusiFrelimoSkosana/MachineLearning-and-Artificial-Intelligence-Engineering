```python
# -*- coding: utf-8 -*-
"""
Created on Sat Aug 24 10:44:42 2022
@author: Vusi Skosana

"""

import pandas as pd

import matplotlib.pyplot as plt

from easyAI.linear_model import gradient
from easyAI.linear_model import predict
from easyAI.linear_model import intercept
from easyAI.linear_model import variance_fit
from easyAI.linear_model import r2_score


df=pd.read_csv('Salary.csv')


#The x,y are in the form of a list although they are not printed as such
x= df['YearsExperience']
y=df['Salary']

m=gradient(x,y)         #gradient/ slope
pred=predict(5.5,x,y)   #predicted value
b=intercept(x,y)        #y-intercept
var= variance_fit(x,y)  #variance arounf the fit
r2= r2_score(x,y)       #r2 score

print('The gradient od the slope is: ',m)
print('The predicted value is: ',pred)
print('The y-intercept is: ',b)
print('variance around the fit is: ',var)
print('The r2 score is ', r2)

y_fit= m*x + b

plt.scatter(x,y, label='Scatter plot')
plt.xlabel("Years of Experience")
plt.ylabel('Salary')
plt.title('Salary vs Years of Experience')
plt.plot(x, y_fit, color='red', label='best fit line')
plt.legend()
plt.show()
```

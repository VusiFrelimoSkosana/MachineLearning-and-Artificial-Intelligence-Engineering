```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
#sAMPLE EMAIL   DATASET

emails = [
    "Buy our new product, it's amazing!",
    "Hello, how are you?",
    "Get a special discount on your next purchase!",
    "Meeting tomorrow at 2 PM.",
    "Limited time offer, don't miss out!"
]
  
labels=[1,0,1,0,1]

vectorizer = CountVectorizer()
X= vectorizer.fit_transform(emails)

classifier =MultinomialNB()
classifier.fit(X, labels)


new_emails= [
    "Congratulations, you won a free vacation!",
    "Reminder: team meeting today at 3 PM"
]

new_X = vectorizer.transform(new_emails)
predictions = classifier.predict(new_X)

for email, prediction in zip(new_emails, predictions):
    print(f"Email: {email}")
    if prediction==1:
        print("Email is spam")
    else:
        print("Email is not spam")
    print()

```

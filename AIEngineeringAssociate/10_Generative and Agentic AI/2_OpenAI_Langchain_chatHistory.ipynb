````python
#Lesson 3
#Make it interactive (like chat)

from openai import OpenAI

client=OpenAI()

while True:
    user_input=input("user: ")

    if user_input in ['quit','exit']:
        break

    else:
        response=client.chat.completions.create(
            model="gpt-4.1",
            messages=[{'role':'user','content':user_input}]
        )

    print(response.choices[0].message.content)
````

````python
#Lesson 4
#(Store Memory)

from openai import OpenAI

client=OpenAI()


conversation=[]  #store the conversation

while True:

    user_input=input("user: ")

    conversation.append({'role':'user', 'content':user_input})

    if user_input.lower() in ['quit','exit']:
        break

    else:
        reply=client.chat.completions.create(
            model='gpt-4.1',
            messages=conversation
        )
        
    assistant_reply= reply.choices[0].message.content
    
    print(assistant_reply)

    conversation.append({'role':'assistant','content':assistant_reply})
````

````python
#Lesson 3 (add Memory to the converation)

from langchain_openai import ChatOpenAI
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory

llm=ChatOpenAI(model='gpt-4.1')

#create the history
history=InMemoryChatMessageHistory()

def get_history():
    return history

#wrap the LLM with memory
chat= RunnableWithMessageHistory(llm, get_session_history=get_history)

response=chat.invoke('what is a lake?')
print('answer: ',response.content)
````

````python
#lesson 4 (combine prompts and history)

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.chat_history import InMemoryChatMessageHistory
from pydantic import BaseModel
from typing import List, Dict
import json

# Manually create DataFrame
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Reference": ["PMDS57838", "PMDS57837", "PMDS57836", "PMDS57837"],
    "Age": [25, 30, 35, 40],
    "Outstanding_fees": [50000, 60000, 70000, 80000],
    "Degree_level": ["NQF 9", "NQF 7", "NQF 10", "NQF 9"],
    "Commnents": ["average less than 65%, already obtained nqf level 9", "NSFAS not attached, In progressed academic record", "applied before july 2024", "satisfied all requirements"],
})


#initiating LLM
llm = ChatOpenAI(model='gpt-4.1-mini')


#creating prompt
prompt = ChatPromptTemplate.from_template(
"""You are an assistant.

A pandas DataFrame {related_topic} is called df already exists.

User question:
{topic}

Return ONLY a valid Python dictionary (not JSON string, not markdown).

The dictionary must have EXACTLY these keys:
- "text"
- "dataframe_code"
- "plot_code"

Format:
{{
  "text": "...",
  "dataframe_code": "...",
  "plot_code": "..."
}}

Rules:

TEXT:
- Only plain explanation
- No code


DATAFRAME_CODE:
- ONLY write pandas code that USES the existing dataframe df
- DO NOT recreate df
- DO NOT define data = [...]
- DO NOT import pandas
- MUST create a variable called result_df
- Use pd.concat() instead of df.append(), as pandas 3.x removed append()


PLOT_CODE:
- ONLY matplotlib code
- show the plot (plt.show())
- Use plt (assume matplotlib.pyplot as plt is already imported)
- Use df or result_df
- DO NOT import matplotlib
- DO NOT show explanations

GLOBAL RULES:
- If dataframe is not needed or user not asked for it → "dataframe_code": "None" , otherwise plot is asked for
- If plot is not needed or not asked for → "plot_code": "None"
- Do NOT change key names
- Do NOT add extra keys
- Output must be directly usable as a Python dictionary
- If the user question does not require a DataFrame, you must literally return "dataframe_code": "None".
- Do not generate any DataFrame code under any circumstances in that case otherwise plots or charts are requested.
""")

#- Do NOT include ``` or markdown
#- No markdown

#Message history
history= InMemoryChatMessageHistory()

def get_history():
    return history
    
#wrap the LLM with history
chat= RunnableWithMessageHistory(llm, get_session_history=get_history)


while True:
    user_input=input('user: ')

    if user_input.lower() in ['quit','exit']:
        break

    else:
        
        #variables
        topic_variables={
            'topic':user_input,
            'related_topic': df.to_json(orient="records", indent=2)
        }
        
        #format prompt
        
        text=prompt.format(**topic_variables)
        
        respond= chat.invoke(text)
        
        answer=respond.content

        data_output = llm.invoke(f"""
        You must strictly decide whether a dataframe/table is NECESSARY to answer the user query.
        
        Rules:
        - Respond with "required" ONLY if the user explicitly asks to see, list, display, or inspect data in a table or dataframe format.
        - Respond with "None" if the question can be answered with explanation, summary, or a plot WITHOUT showing a dataframe.
        - Do NOT assume a dataframe is needed.
        - Be conservative: if unsure, respond with "None".
        - 
        
        Output format:
        - Respond with ONLY one word: "required" or "None"
        - Do NOT include any explanation or extra text.
        
        User query:
        {user_input}
        """)
        
        data_corrector=data_output.content
        #print(data_corrector)

        plot_output = llm.invoke(f"""
        You must strictly decide whether a plot/charts is NECESSARY to answer the user query.
        
        Rules:
        - Respond with "required" ONLY if the user explicitly asks to see any plots or charts.
        - Respond with "None" if the question can be answered with explanation, summary, or a data WITHOUT showing a plots.
        - Do NOT assume a plot is needed.
        - Be conservative: if unsure, respond with "None".
        
        Output format:
        - Respond with ONLY one word: "required" or "None"
        - Do NOT include any explanation or extra text.
        
        User query:
        {user_input}
        """)

        plot_corrector=plot_output.content
        #print(plot_corrector)

        #We need to ensure that the answer is a dictionary 
        final_answer= json.loads(answer)

        #print(final_answer)
        
        text=final_answer['text']
        print("\n Assistant: \n"+str(text))
        
        dataframe_code=final_answer['dataframe_code']
        plot_code=final_answer['plot_code']

        #create local vars
        glob_vars={"df": df, "pd":pd}
        # Evaluate the code to get the resulting DataFrame
        exec(dataframe_code, glob_vars)
        result_df= glob_vars.get("result_df")

        if data_corrector != "None":
            try:
                display(result_df)
                
            except Exception as e:
                print("Error executing dataframe code:", e)

        
        
        if plot_corrector != "None":
            try:
                #add local variable
                local_vars={"df": df, "plt": plt, "np":np,"result_df": result_df}
                # Execute the plotting code with df and plt in scope
                exec(plot_code, {},local_vars )
            except Exception as e:
                print("Error executing plot code:", e)

    
````
